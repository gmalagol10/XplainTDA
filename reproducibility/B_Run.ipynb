```python
import os
```

# Stability

## kNN

```python
for dataset, label in zip(["PBMC","HumanBrain","BMMCMultiOme"], ["Raw_CellType", "Raw_celltype","Coarse_CellType"]):
    for omic in ["GEX","Peak"]:
        file=f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad"
        if os.path.isfile(file):
            print(dataset, label, omic)
            os.system(f"sbatch -p slim18 -J stb{dataset[:2]}{omic} --mem=200G --ntasks=1 --cpus-per-task=16 --time 7-00:00:00 --exclusive\
               --output=Datasets/{dataset}/{dataset}_{omic}_Run_Stability_kNN_GRAE.out Run_Exp.sh Run_Stability_kNN \
               Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad Datasets/{dataset}/XplainTDA/{omic}/Stability/GRAE/kNN X_GRAE")
```

## Subsampling

```python
for dataset, label in zip(["PBMC","HumanBrain","BMMCMultiOme"], ["Raw_CellType", "Raw_celltype","Coarse_CellType"]):
    for omic in ["GEX","Peak"]:
        file=f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad"
        if os.path.isfile(file):
            print(dataset, label, omic)
            os.system(f"sbatch -p slim18 -J sub{dataset[:2]}{omic} --mem=32G --ntasks=1 --cpus-per-task=6 --time 7-00:00:00\
               --output=Datasets/{dataset}/{dataset}_{omic}_Run_Stability_Subsampling_GRAE.out Run_Exp.sh Run_Stability_Subsampling \
               Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad Datasets/{dataset}/XplainTDA/{omic}/Stability/GRAE/Subsampling \
               --stratify-by {label} --embedding-key X_GRAE --n-neighbors 75")
```

# Annotation 

```python
omic="GEX"
for dataset, label in zip(["PBMC","HumanBrain","BMMCMultiOme"], ["Raw_CellType", "Raw_celltype","Coarse_CellType"]):
    file=f"Datasets/{dataset}/CM/{dataset}_GEX_Def.h5ad"
    if os.path.isfile(file):
        print(dataset, label, omic)
        os.system(
        f"sbatch -p slim18 "
        f"-J tdaAnn{dataset} "
        f"--mem=40G --ntasks=1 --cpus-per-task=12 "
        f"--time=7-00:00:00 "
        f"--output=Datasets/{dataset}/{dataset}_GEX_{label}_Run_Annotation.out "
        f"./Run_Exp.sh Run_Annotation "
        f"--adata-path {file} "
        f"--celltype-col {label} "
        f"--out-path Datasets/{dataset}/XplainTDA/{omic}/Annotation/{label}")
```

# Explainer

```python
for dataset, label in zip(["PBMC","HumanBrain","BMMCMultiOme"], ["Raw_CellType", "Raw_celltype","Coarse_CellType"]):
    for omic in ["GEX","Peak"]:
        file=f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad"
        if os.path.isfile(file):
            print(dataset, label, omic)
            os.system(
            f"sbatch -p slim18 "
            f"-J xTD{dataset[:2]}{omic} "
            f"--mem=128G --ntasks=1 --cpus-per-task=30 "
            f"--time=7-00:00:00 --exclusive "
            f"--output=Datasets/{dataset}/{dataset}_{omic}_{label}_Run_Stratified_Explanations_GRAE.out "
            f"Run_Exp.sh Run_Stratified_Explanations "
            f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad "
            f"Datasets/{dataset}/XplainTDA/{omic}/Explanations/{label} "
            f"{label} "
            f"--embedding-key X_GRAE "
            f"--grae-model-path Datasets/{dataset}/CM/{dataset}_{omic}_GRAE_embedder.joblib")
```
