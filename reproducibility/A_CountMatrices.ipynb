```python
import numpy as np
import scanpy as sc
import pandas as pd
import matplotlib.pyplot as plt
import joblib
import Embedder
import os
import Utils as ut
```

```python
for dataset, label in zip(["PBMC","HumanBrain","BMMCMultiOme"], ["Raw_CellType", "Raw_celltype","Coarse_CellType"]):
    for omic in ["GEX","Peak"]:
        file=f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad"
        if os.path.isfile(file):
            print(dataset, label, omic)
            adata = sc.read_h5ad(f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad")       
            labels = adata.obs[label].to_numpy()
            embedder_cls = Embedder.Embedder.registry["GRAE"]
            embedder = embedder_cls(n_components=int(np.round(adata.shape[1] ** (1 / 3))), stratify_labels=labels)
            Z = embedder.fit_transform(adata.X)
            adata.obsm["X_GRAE"]=Z
            adata.write(f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad", compression="gzip")
            embedder.save(f"Datasets/{dataset}/CM/{dataset}_{omic}_GRAE_embedder.joblib")
```

```python
for dataset, label in zip(["PBMC","HumanBrain","BMMCMultiOme"], ["Raw_CellType", "Raw_celltype","Coarse_CellType"]):
    for omic in ["GEX"]:
        file=f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad"
        adata = sc.read_h5ad(file)
        print(dataset, adata.shape)
```
