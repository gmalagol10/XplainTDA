# PBMC

## GEX

```python
adata=sc.read_h5ad("ADATA_PATH_TO_THE_RAW_COUNT_MATRIX")
adata
```

```python
adata=ut.qc_filtering(adata, omic="GEX")
```

```python
adata
```

```python
# create coarse cell-type annotation
adata.obs["Raw_CellType"] = pd.Series(index=adata.obs.index, dtype="object")

adata.obs.loc[adata.obs["CellType"].str.contains(r"^B_", regex=True), "Raw_CellType"] = "B"
adata.obs.loc[adata.obs["CellType"].str.contains(r"^T_", regex=True),"Raw_CellType"] = "T"
adata.obs.loc[adata.obs["CellType"].str.contains(r"^NK$", regex=True), "Raw_CellType"] = "NK"
adata.obs.loc[adata.obs["CellType"].str.contains(r"^Monocytes$", regex=True),"Raw_CellType"] = "Monocytes"
adata.obs.loc[adata.obs["CellType"].str.contains(r"^DC", regex=True),"Raw_CellType"] = "DC"

# optional: remove cells not assigned
adata = adata[adata.obs["Raw_CellType"].notna()].copy()

# check result
adata.obs["Raw_CellType"].value_counts()
```

```python
adata
```

```python
adata.write("Datasets/PBMC/CM/PBMC_GEX_Def.h5ad", compression="gzip")
```

# BMMCMultiOme

## GEX

```python
adata=sc.read_h5ad("ADATA_PATH_TO_THE_RAW_COUNT_MATRIX")
adata
```

```python
coarse_map = {
    "CD4+ T activated": "T",
    "CD4+ T naive": "T",
    "CD8+ T": "T",

    "Naive CD20+ B": "B",
    "B1 B": "B",
    "Transitional B": "B",
    "Plasma cell": "B",

    "NK": "NK/ILC",
    "ILC": "NK/ILC",

    "CD14+ Mono": "Monocyte",
    "CD16+ Mono": "Monocyte",

    "pDC": "DC",
    "cDC2": "DC",

    "Erythroblast": "Erythroid",
    "Normoblast": "Erythroid",
    "Proerythroblast": "Erythroid",

    "HSC": "Progenitor",
    "Lymph prog": "Progenitor",
    "MK/E prog": "Progenitor",
    "G/M prog": "Progenitor",
    "ID2-hi myeloid prog": "Progenitor",
}

adata.obs["Coarse_CellType"] = adata.obs["CellType"].map(coarse_map)

adata.obs["Coarse_CellType"].value_counts(dropna=False)
```

```python
adata.write("Datasets/BMMCMultiOme/CM/BMMCMultiOme_GEX_Def.h5ad", compression="gzip")
```

# GRAE

```python
for dataset, label in zip(["PBMC","HumanBrain","BMMCMultiOme"], ["Raw_CellType", "Raw_celltype","Coarse_CellType"]):
    for omic in ["GEX","Peak"]:
        file=f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad"
        if os.path.isfile(file):
            print(dataset, label, omic)
            adata = sc.read_h5ad(f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad")       
            labels = adata.obs[label].to_numpy()
            embedder_cls = Embedder.Embedder.registry["GRAE"]
            embedder = embedder_cls(n_components=int(np.round(adata.shape[1] ** (1 / 3))), stratify_labels=labels)
            Z = embedder.fit_transform(adata.X)
            adata.obsm["X_GRAE"]=Z
            adata.write(f"Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad", compression="gzip")
            embedder.save(f"Datasets/{dataset}/CM/{dataset}_{omic}_GRAE_embedder.joblib")
```
