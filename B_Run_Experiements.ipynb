```python
for dataset, label in zip(["PBMC","HumanBrain","MouseBrain"], ["Raw_CellType", "Raw_celltype", "CellType"]):
    for omic in ["GEX","Peak"]:
        os.system(f"sbatch -p slim18 -J xTDAG{dataset[:2]}{omic} --mem=180G --ntasks=1 --cpus-per-task=18 --time 7-00:00:00 --exclusive\
               --output=Datasets/{dataset}/{dataset}_{omic}_{label}_Run_Stratified_Explanations_GRAE.out Run_Exp.sh Run_Stratified_Explanations \
               Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad Datasets/{dataset}/XplainTDA/{omic}/Explanations/{label} {label} --embedding-key X_GRAE  --grae-model-path Datasets/{dataset}/CM/{dataset}_{omic}_GRAE_embedder.joblib")
```
