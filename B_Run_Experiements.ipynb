## Subsampling

```python
for dataset, label in zip(["PBMC","HumanBrain","MouseBrain"], ["Raw_CellType", "Raw_celltype", "CellType"]):
    for omic in ["GEX","Peak"]:
        print(dataset, omic)
        os.system(f"sbatch -p slim18 -J sub{dataset[:2]}{omic} --mem=40G --ntasks=1 --cpus-per-task=8 --time 7-00:00:00\
               --output=Datasets/{dataset}/{dataset}_{omic}_Run_Stability_Subsampling.out Run_Exp.sh Run_Stability_Subsampling \
               Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad Datasets/{dataset}/XplainTDA/{omic}/Stability/Subsampling {label}")
```
