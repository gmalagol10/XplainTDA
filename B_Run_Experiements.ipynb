# Stability

## kNN

```python
for dataset in ["PBMC","HumanBrain","MouseBrain"]:
    for omic in ["GEX","Peak"]:
        print(dataset, omic)
        os.system(f"sbatch -p slim18 -J stb{dataset[:2]}{omic} --mem=200G --ntasks=1 --cpus-per-task=10 --time 7-00:00:00 --exclusive\
               --output=Datasets/{dataset}/{dataset}_{omic}_Run_Stability_kNN.out Run_Exp.sh Run_Stability_kNN \
               Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad Datasets/{dataset}/XplainTDA/{omic}/Stability/kNN")
```

```python
os.system("./Run_Exp.sh Run_Stability_kNN Datasets/MouseBrain/CM/MouseBrain_GEX_Def.h5ad Datasets/MouseBrain/XplainTDA/GEX/Stability/kNN X_GRAE")
```

## Subsampling

```python
for dataset, label in zip(["PBMC","HumanBrain","MouseBrain"], ["Raw_CellType", "Raw_celltype", "CellType"]):
    for omic in ["GEX","Peak"]:
        print(dataset, omic)
        os.system(f"sbatch -p slim16 -J sub{dataset[:2]}{omic} --mem=32G --ntasks=1 --cpus-per-task=6 --time 7-00:00:00\
               --output=Datasets/{dataset}/{dataset}_{omic}_Run_Stability_Subsampling.out Run_Exp.sh Run_Stability_Subsampling \
               Datasets/{dataset}/CM/{dataset}_{omic}_Def.h5ad Datasets/{dataset}/XplainTDA/{omic}/Stability/Subsampling {label} X_GRAE")
```

```python
os.system("./Run_Exp.sh Run_Stability_Subsampling Datasets/MouseBrain/CM/MouseBrain_GEX_Def.h5ad Datasets/MouseBrain/XplainTDA/GEX/Stability/Subsampling CellType X_GRAE")
```
