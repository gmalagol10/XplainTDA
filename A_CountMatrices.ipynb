# PBMC

## GEX

```python
adata=sc.read_h5ad("../SEAGALL/Datasets/10XhsPBMC10kMO/FeatureSpaces/GEX/CM/10XhsPBMC10kMO_GEX_Raw.h5ad")
adata
```

```python
adata=ut.qc_filtering(adata, omic="GEX")
```

```python
adata
```

```python
# create coarse cell-type annotation
adata.obs["Raw_CellType"] = pd.Series(index=adata.obs.index, dtype="object")

adata.obs.loc[adata.obs["CellType"].str.contains(r"^B_", regex=True), "Raw_CellType"] = "B"
adata.obs.loc[adata.obs["CellType"].str.contains(r"^T_", regex=True),"Raw_CellType"] = "T"
adata.obs.loc[adata.obs["CellType"].str.contains(r"^NK$", regex=True), "Raw_CellType"] = "NK"
adata.obs.loc[adata.obs["CellType"].str.contains(r"^Monocytes$", regex=True),"Raw_CellType"] = "Monocytes"
adata.obs.loc[adata.obs["CellType"].str.contains(r"^DC", regex=True),"Raw_CellType"] = "DC"

# optional: remove cells not assigned
adata = adata[adata.obs["Raw_CellType"].notna()].copy()

# check result
adata.obs["Raw_CellType"].value_counts()
```

```python
adata.write("Datasets/PBMC/CM/PBMC_GEX_Def.h5ad", compression="gzip")
```
